```python
from pyPM5 import PM5
import copy
import numpy as np
import matplotlib.pyplot as plt
import time

pm=PM5()
pm.connect('USB0::1027::24577::856V::0::RAW')
```

```python
pm.set_range(3,1)
```

```python
pm.set_cal_heater(0)
```

```python
count=0
power = []

while(count<100):
    power.append(pm.read_power())
    print(power)
    count+=1
    
```
